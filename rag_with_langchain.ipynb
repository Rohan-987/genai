RAG with langchain

We create 2 important pipelines. 
1.Data ingestion pipeline
2.Retrieval pipeline

1. PARSING 
Parsing = Reading and extracting raw text from your documents.Your data comes in many formats — PDF, HTML, Excel, SQL etc. The LLM can't read a PDF directly. Parsing converts them into plain text.Parsing is like a scanner — it reads whatever format your document is in and converts it to plain readable text.

RAW DATA              AFTER PARSING
─────────────────     ─────────────────────
policy.pdf       →    "Employees get 18 days
                       of paid leave per year..."

data.html        →    "Product price is $99.
                       Available in 3 colors..."

2. EMBEDDING 
Embedding = Converting that plain text into numbers (vectors) that capture meaning
After parsing you have text. But computers/LLMs don't understand text directly — they understand numbers. Embedding converts text into a list of numbers called a vector.Embedding is like giving every piece of text a GPS coordinate. Similar meanings → similar coordinates → close to each other in space.                       

TEXT                        VECTOR (simplified)
──────────────────          ──────────────────────────────
"leave policy"         →    [0.23, 0.81, 0.45, 0.12, ...]
"vacation rules"       →    [0.24, 0.80, 0.44, 0.11, ...]
"company founded 2010" →    [0.91, 0.02, 0.87, 0.63, ...]
